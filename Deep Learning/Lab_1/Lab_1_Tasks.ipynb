# Task 1: Create a NumPy array that contains numbers from 0 to 32, then get the power of 2 for each element.

The output should be like [1, 2, 4, 8, 16, 32 ...]

```python
import numpy as np

numbers = np.array(range(0, 33))

powers = np.power(2, numbers)

print(powers)
```

#  Task 2: Using selectors, crop the pixels in this rectangle (350, 70) to (570, 200), and show the copred image.
Notice from the shape what are the fisrt and second axis.

```python
# 1. تحميل المكتبات والصورة
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import image

# 2. تحميل الصورة
img = image.imread("penguin.jpg")

# 3. الاقتصاص
# تذكر: [صفوف, أعمدة] = [y, x]
y_start = 70    # بداية الصفوف
y_end = 200     # نهاية الصفوف
x_start = 350   # بداية الأعمدة
x_end = 570     # نهاية الأعمدة

cropped_img = img[y_start:y_end, x_start:x_end]

# 4. عرض النتيجة
plt.figure(figsize=(10, 6))

# عرض الصورة الأصلية مع المستطيل
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Full Photo")
plt.gca().add_patch(plt.Rectangle((x_start, y_start), 
                                 x_end-x_start, 
                                 y_end-y_start,
                                 linewidth=3, 
                                 edgecolor='red', 
                                 facecolor='none'))

# عرض الصورة المقتطعة
plt.subplot(1, 2, 2)
plt.imshow(cropped_img)
plt.title(f"Cropped Area\n({cropped_img.shape[1]}×{cropped_img.shape[0]})")

plt.show()
```

# Task 3: Make the penguin anonymous!
Cover the penguin eyes with a black rectangle, to understand how to do this, you should first understand what a color is!
The eyes rectangle is (410, 125) -> (480, 145).
*Note:* Please copy the image using `np.copy`.

```python
# 1. تحميل المكتبات
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import image

# 2. تحميل ونسخ الصورة
penguin = image.imread("penguin.jpg")
anonymous = np.copy(penguin)

# 3. رسم المستطيل الأسود
anonymous[125:145, 410:480] = [0, 0, 0]

# 4. عرض بسيط مع المحاور
plt.figure(figsize=(14, 6))

# الصورة الأصلية
plt.subplot(1, 2, 1)
plt.imshow(penguin)
plt.title("original Photo", fontsize=14)
plt.xlabel("X", fontsize=12)
plt.ylabel("Y", fontsize=12)


# الصورة المعدلة
plt.subplot(1, 2, 2)
plt.imshow(anonymous)
plt.title("After Edits", fontsize=14)
plt.xlabel("X", fontsize=12)
plt.ylabel("Y", fontsize=12)

# رسم المستطيل الأسود
plt.gca().add_patch(plt.Rectangle((410, 125), 70, 20, 
                                 linewidth=2, edgecolor='red', 
                                 facecolor='black', alpha=0.3))

plt.tight_layout()
plt.show()
```

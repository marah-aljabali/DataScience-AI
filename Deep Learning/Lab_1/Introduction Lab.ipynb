One of the most important things about NumPy arrays is that you could do mathematical operations directly on them (without using explicit loops), which will make your code runs way faster.

```python
array_10000 = np.array(range(0,10000)) #NumPy array ---> [0,1,2,3,4,....,9999]
print("Array Length: "+str(len(array_10000)))
```

```python
# calculating list Product using numpy

%%time  # This will print the execution time
array_10000_mul_5 = array_10000 * 5           
print (array_10000_mul_5[2]) # 2 * 5 = 10
```

```python
# calculating list Product using for loop

%%time
list_10000 = range(0,10000) #Python array ---> [0,1,2,3,4,....,9999]
list_10000_mul_5 = []

for i in range(len(list_10000)):
  list_10000_mul_5.append(list_10000[i] * 5)
print (list_10000_mul_5[2]) # 2 * 5 = 10
```

#### What is a pixel?
Depending on the color space of the image, the pixel value could be a single number that indicates the brightness, or a multiple colors brightness (e.g. Red, Green, and Blue), the value could be a byte for each color (e.g. from 0 -> 255), but this could vary depending on the color space (e.g most game engines and GPUs uses floating-point numbers).

```python
print(penguin[0,0]) # The value of the fisrt pixel
```

To create a white image, we could create a `100px` X `100px` array, in RGB space with 255 value for each color.

```python
all_white_image = np.full((100,100,3), 255)
plt.imshow(all_white_image)
plt.show()
```

A black image, we be the same, but the values for all colors will be zeros.

```python
all_black_image = np.full((100,100,3), 0)
plt.imshow(all_black_image)
plt.show()
```

To create a red image, we will write 255 to the first element in each pixel.

```python
all_red_image = np.copy(all_black_image)
all_red_image[:,:,0] = 255  # Set the red channel (first column) to 255 (R G B)
# if we want to make green use all_red_image[:,:,1] = 255
# if we want to make blue use all_red_image[:,:,2] = 255
plt.imshow(all_red_image)
plt.show()
```

### Vectors

A vector is an object that has both a magnitude and a direction. Geometrically, we can picture a vector as a directed line segment, whose length is the magnitude of the vector and with an arrow indicating the direction. The direction of the vector is from its tail to its head. [[1]](https://mathinsight.org/vector_introduction)


$Vector = [x, y]$

So to get the magnitude of the vector, we should get the square root of the summation of all its components squared.

$Magnitude = |Vector| = \sqrt{x^2+y^2}$

$Angle = tan^{-1}(y/x)$

For example: 

$V_1 = [4, 3] \\
|V_1| = \sqrt{16 + 9} = 5
$

We can get the angle of this vector by getting the $tan^{-1}$ of its x, y components:

$
tan^{-1}({{3}\over {4}}) = 36.8^{\circ}
$

```python
plt.arrow(0,0,4,3, width=0.05, length_includes_head=True, fc="black")

plt.xlim(0,5)
plt.ylim(0,5)
plt.grid()
plt.show()
```

#### Numpy and Vectors

Numpy has a lot of methods to manipulate vectors, for example, you can use `.dot(v2)` to get the dot product between two vectors.

$V_1 = [x_1, y_1]\\
V_2 = [x_2, y_2]\\
V_1 . V_2 = |V_1| * |V_2| * cos(\theta) = x_1 * x_2 + y_1 * y_2
$

```python
v_1 = np.array([5, 10, 15])
v_2 = np.array([0.5, 0.5, 0])
print("v1 dot v2 = "+str(v_1.dot(v_2)))
```

#### Dot product with colors!

```python
flower = image.imread("flower.jpg")
selectedPixel = flower[800, 100]

plt.plot(100, 800, 'rx')
plt.imshow(flower)
plt.show()

print("Pixel at 100, 800   = "+str(selectedPixel))
v = np.array([0,0,1])

print("Pixel dot [0, 0, 1] = "+str(selectedPixel.dot(v)))
```

Notice the pixel is marked with a red cross.

```python
red_channel = np.zeros(flower.shape, dtype='u1')
green_channel = np.zeros(flower.shape, dtype='u1')
blue_channel = np.zeros(flower.shape, dtype='u1')

red_channel[:,:,0] = flower[:,:].dot(np.array([1,0,0]))
green_channel[:,:,1] = flower[:,:].dot(np.array([0,1,0]))
blue_channel[:,:,2] = flower[:,:].dot(np.array([0,0,1]))

plt.imshow(red_channel)
plt.show()
plt.imshow(green_channel)
plt.show()
plt.imshow(blue_channel)
plt.show()
```

### More into numpy
There are a lot of useful methods in numpy, which can help you mainpulating matrices and images, for example, you could get the mean of each pixel (mean color) simply by writing `mean(axis=2)`

```python
mean = penguin.mean(axis=2)
show = plt.imshow(mean)
plt.colorbar(show)
plt.show()
```

Or to show it as gray scale, we could tell matplotlib to use grayscale!

```python
show = plt.imshow(mean, cmap='gray')
plt.colorbar(show)
plt.show()
```

So, to actualy convert the image to gray scale, you will give weight for each of the colors compoenet

$Grayscale = 0.299R + 0.587G + 0.114B$

You can use dot product to easily achive this

```python
gray_weight = np.array([0.299, 0.587, 0.114])
gray = penguin[:,:].dot(gray_weight)
show = plt.imshow(gray, cmap='gray')
plt.colorbar(show)
plt.show()
```
